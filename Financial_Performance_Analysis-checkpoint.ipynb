```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# 1. Load Datasets
df_budget = pd.read_csv('Budget.csv')
df_customers = pd.read_csv('Customers.csv')
df_trans = pd.read_csv('Financial_Transactions.csv')
df_headcount = pd.read_csv('Headcount.csv')
df_vendors = pd.read_csv('Vendors.csv')

# 2. Fix Date Columns (Crucial Step: use dayfirst=True)
df_trans['transaction_date'] = pd.to_datetime(df_trans['transaction_date'], dayfirst=True)
df_customers['join_date'] = pd.to_datetime(df_customers['join_date'], dayfirst=True)
df_headcount['join_date'] = pd.to_datetime(df_headcount['join_date'], dayfirst=True)

# 3. Create a helper column for Monthly Analysis
df_trans['month_year'] = df_trans['transaction_date'].dt.to_period('M')

print("🚀 Environment Ready! All 5 tables loaded and dates synchronized.")
print(f"Total Transactions: {len(df_trans)}")
```

```python
# Separate Revenue and Expenses
revenue_data = df_trans[df_trans['account_type'] == 'Revenue']['amount']
expense_data = df_trans[df_trans['account_type'] == 'Expense']['amount']

# Calculate Measures of Central Tendency & Dispersion
def get_stats(data, title):
    print(f"--- {title} Statistics ---")
    print(f"Mean:   ${data.mean():,.2f}")
    print(f"Median: ${data.median():,.2f}")
    print(f"Std Dev: ${data.std():,.2f}")
    print(f"Range:  ${data.min():,.2f} to ${data.max():,.2f}\n")

get_stats(revenue_data, "Revenue")
get_stats(expense_data, "Expense")
```

```python
# --- STEP 4: STATISTICAL ANALYSIS ---

# 1. ANOVA: Does Mean Monthly Expense differ significantly across Business Units?
# First, aggregate expenses by month and unit
df_trans['month'] = df_trans['transaction_date'].dt.month
df_trans['year'] = df_trans['transaction_date'].dt.year

monthly_expenses = df_trans[df_trans['account_type'] == 'Expense'].groupby(
    ['year', 'month', 'business_unit'])['amount'].sum().abs().reset_index()

# Separate groups for ANOVA
grps = [monthly_expenses[monthly_expenses['business_unit'] == unit]['amount'] 
        for unit in monthly_expenses['business_unit'].unique()]

f_stat, p_anova = stats.f_oneway(*grps)

print("--- 🔬 ANOVA RESULT: Expense by Business Unit ---")
print(f"F-Statistic: {f_stat:.4f}, P-Value: {p_anova:.4e}")
if p_anova < 0.05:
    print("Interpretation: ✅ SIGNIFICANT. Business units have fundamentally different spending scales.\n")
else:
    print("Interpretation: ❌ NOT SIGNIFICANT. Differences are likely due to random chance.\n")


# 2. T-TEST: Compare Average Revenue between North and West regions
north_rev = df_trans[(df_trans['account_type'] == 'Revenue') & (df_trans['region'] == 'North')]['amount']
west_rev = df_trans[(df_trans['account_type'] == 'Revenue') & (df_trans['region'] == 'West')]['amount']

t_stat, p_ttest = stats.ttest_ind(north_rev, west_rev)

print("--- 🔬 T-TEST RESULT: North vs West Revenue ---")
print(f"T-Statistic: {t_stat:.4f}, P-Value: {p_ttest:.4f}")
if p_ttest < 0.05:
    print("Interpretation: ✅ SIGNIFICANT. There is a real performance gap between North and West.\n")
else:
    print("Interpretation: ❌ NOT SIGNIFICANT. These regions perform similarly on average.\n")


# 3. CHI-SQUARE: Customer Segment vs. Over-Budget Status
# First, define 'Over-Budget' at the Business Unit/Month level
actual_exp = df_trans[df_trans['account_type'] == 'Expense'].groupby(['year', 'month', 'business_unit'])['amount'].sum().abs().reset_index()
actual_exp.columns = ['year', 'month', 'business_unit', 'Actual']

budget_status = pd.merge(actual_exp, df_budget, on=['year', 'month', 'business_unit'])
budget_status['is_over_budget'] = budget_status['Actual'] > budget_status['budgeted_expense']

# Merge status back to transactions to link with Customer Segments
df_merged = pd.merge(df_trans, budget_status[['year', 'month', 'business_unit', 'is_over_budget']], on=['year', 'month', 'business_unit'])
df_with_cust = pd.merge(df_merged, df_customers[['customer_id', 'segment']], on='customer_id')

# Create Contingency Table
contingency = pd.crosstab(df_with_cust['segment'], df_with_cust['is_over_budget'])
chi2, p_chi, dof, ex = stats.chi2_contingency(contingency)

print("--- 🔬 CHI-SQUARE RESULT: Segment vs. Budget Status ---")
print(f"Chi-Square: {chi2:.4f}, P-Value: {p_chi:.4f}")
if p_chi < 0.05:
    print("Interpretation: ✅ SIGNIFICANT. Certain customer segments are linked to months of over-spending.\n")
else:
    print("Interpretation: ❌ NOT SIGNIFICANT. Customer segment doesn't predict budget overruns.\n")
```

```python
# --- STEP 5: BUDGET VARIANCE & COST ANALYSIS ---

# 1. Aggregate Transactions to match Budget granularity (Year, Month, Unit)
expenses = df_trans[df_trans['account_type'] == 'Expense']
revenues = df_trans[df_trans['account_type'] == 'Revenue']

actual_exp = expenses.groupby(['year', 'month', 'business_unit'])['amount'].sum().abs().reset_index()
actual_exp.columns = ['year', 'month', 'business_unit', 'Actual_Expense']

# 2. Merge with Budget table
df_variance = pd.merge(df_budget, actual_exp, on=['year', 'month', 'business_unit'], how='inner')

# 3. Calculate Variance & Utilization
df_variance['Exp_Variance'] = df_variance['Actual_Expense'] - df_variance['budgeted_expense']
df_variance['Utilization_Pct'] = (df_variance['Actual_Expense'] / df_variance['budgeted_expense']) * 100

# 4. Break down actual expenses by category to locate the spikes
cat_breakdown = expenses.groupby(['year', 'month', 'business_unit', 'category'])['amount'].sum().abs().reset_index()

print("--- 💸 Top 5 Biggest Budget Overruns ---")
display(df_variance.sort_values(by='Exp_Variance', ascending=False).head())

print("\n--- 📉 Top 5 Most Underutilized Budgets ---")
display(df_variance.sort_values(by='Utilization_Pct', ascending=True).head())
```

```python
# Chart 1: Average Monthly Expense Variance by Business Unit
plt.figure(figsize=(12, 6))
sns.barplot(data=df_variance, x='business_unit', y='Exp_Variance', palette='coolwarm', estimator=np.mean)
plt.axhline(0, color='black', linestyle='--')
plt.title('Average Monthly Expense Variance by Business Unit (Positive = Over Budget)', fontsize=14)
plt.ylabel('Variance Amount ($)')
plt.show()

# Chart 2: Average Expense by Category (To find Cost Spikes)
plt.figure(figsize=(12, 6))
sns.barplot(data=cat_breakdown, x='category', y='amount', palette='viridis', estimator=np.mean)
plt.title('Average Expense by Category to Detect Spikes', fontsize=14)
plt.ylabel('Amount ($)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# --- STEP 6: HEADCOUNT & ROI ANALYSIS ---

# 1. Calculate Total Revenue by Business Unit
bu_revenue = df_trans[df_trans['account_type'] == 'Revenue'].groupby('business_unit')['amount'].sum().reset_index()
bu_revenue.columns = ['business_unit', 'Total_Revenue']

# 2. Calculate Headcount and Total Employee Cost from Headcount table
bu_headcount = df_headcount.groupby('business_unit').agg(
    Employee_Count=('employee_id', 'count'),
    Total_Payroll_Cost=('cost_to_company', 'sum')
).reset_index()

# 3. Merge for ROI Analysis
df_roi = pd.merge(bu_revenue, bu_headcount, on='business_unit')

# 4. Calculate Key Performance Indicators (KPIs)
df_roi['Revenue_per_Employee'] = df_roi['Total_Revenue'] / df_roi['Employee_Count']
df_roi['ROI_Ratio'] = df_roi['Total_Revenue'] / df_roi['Total_Payroll_Cost']

print("--- 👤 Employee Efficiency & ROI Summary ---")
display(df_roi.sort_values(by='Revenue_per_Employee', ascending=False))

# Visualization: Revenue per Employee
plt.figure(figsize=(10, 6))
sns.barplot(data=df_roi, x='business_unit', y='Revenue_per_Employee', palette='magma')
plt.title('Productivity: Revenue Generated per Employee by Unit', fontsize=14)
plt.ylabel('Revenue per Employee ($)')
plt.show()
```

```python
# Check Vendor Concentration (Corrected)
# We join on vendor_id but keep the transaction 'amount'
vendor_spend = df_trans[df_trans['account_type'] == 'Expense'].merge(df_vendors, on='vendor_id', how='left')

# Grouping by vendor_name and summing 'amount'
top_vendors = vendor_spend.groupby('vendor_name')['amount'].sum().abs().sort_values(ascending=False)

print("--- 💸 Top 10 Vendors by Spend ---")
print(top_vendors.head(10))
```

```python
# 1. Replace NaNs in specific columns to keep the data clean
# For Transactions
df_trans['customer_id'] = df_trans['customer_id'].fillna('Internal/None')
df_trans['vendor_id'] = df_trans['vendor_id'].fillna('Internal/None')

# For Customers and Vendors (if any exist)
df_customers = df_customers.fillna('N/A')
df_vendors = df_vendors.fillna('N/A')

# 2. Re-create the Master Merge (Ensuring we use correct column names)
df_pbi_master = df_trans.copy()

# Join Customer segment info
df_pbi_master = df_pbi_master.merge(df_customers[['customer_id', 'segment', 'status']], 
                                    on='customer_id', 
                                    how='left')

# Join Vendor name and their category
# We use suffixes because both Transactions and Vendors have a 'category' column
df_pbi_master = df_pbi_master.merge(df_vendors[['vendor_id', 'vendor_name', 'category']], 
                                    on='vendor_id', 
                                    how='left', 
                                    suffixes=('', '_vendor'))

# 3. Final Fill for any 'orphaned' records after merging
df_pbi_master['segment'] = df_pbi_master['segment'].fillna('N/A')
df_pbi_master['vendor_name'] = df_pbi_master['vendor_name'].fillna('N/A')
df_pbi_master['category_vendor'] = df_pbi_master['category_vendor'].fillna('Direct Revenue')

# 4. Export the files
df_pbi_master.to_csv('Master_Fact_Table.csv', index=False)
df_variance.to_csv('Budget_Variance_Table.csv', index=False)

print("✅ Master files are saved on the server. Ready for download!")
```
